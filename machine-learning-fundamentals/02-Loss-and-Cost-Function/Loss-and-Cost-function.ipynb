## MSE (Mean Squared Error)

In this exercise, we will be computing the cost function for the predictions of house prices based on the given data.
The only attribute in out data is the "size in 1000 sqft" and the "target 1000s of $" is the price. I have experimented with different values of w and b, and these values seem to be working better.

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
w = 200
b = 50
```

```python
def cost(x, y, w, b): 

    m = x.shape[0] 
    
    cost_sum = 0 #agregated cost
    for i in range(m): 
        f = w * x[i] + b   #predictions
        cost = (f - y[i]) ** 2  #cost of an individual instance
        cost_sum = cost_sum + cost  
    total_cost = (1 / (2 * m)) * cost_sum  #cost function

    return total_cost
```

```python
x_train = np.array([1.0, 1.7, 2.0, 2.5, 3.0, 3.2])
y_train = np.array([250, 300, 480,  430,   630, 730,])
f_data = [w * x_train[i] + b for i in range(x_train.shape[0])]
print(cost(x_train, y_train, w, b))
```

The following cell is the visualization of out model's behaviour. You can alter the value of w and b to observe the changes in its behaviour.

```python
# Plot the data points
plt.scatter(x_train, y_train, marker='o', c='r')
plt.plot(x_train, f_data, c='b',label='model prediction')
# Set the title
plt.title("Housing Price")
# Set the y-axis label
plt.ylabel('Price (in 1000 $)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.show()

```

## Logistic Cost Function

In the last section we implemented mse loss and cost function which works well for linear regression algorithm. Logistic loss fucntion is well suited for regression algorithm including *sigmoid* as *activation function*. We will study activation functions later.

```python
def sigmoid(z): #z is the output of the linear equation f(w,b)
    s = 1/(1+np.exp(-z)) #np.exp is the Euler's number: e
    return s



def logistic_cost(X, y, w, b):

    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        z_i = np.dot(X[i],w) + b
        f_wb_i = sigmoid(z_i)
        cost +=  -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1-f_wb_i) #logistic loss
             
    cost = cost / m
    return cost
```

```python
x_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])  #(m,n)
y_train = np.array([0, 0, 0, 1, 1, 1])                                           #(m,)
w = np.array([1,1])
b = -3
print(logistic_cost(x_train, y_train, w, b))
```
